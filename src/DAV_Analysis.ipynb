```python
# ============================================================
# DAV PROJECT
# Explainable Data Analytics for Sustainable Smart City Applications
# STEP 1: DATASET LOADING & INITIAL INSPECTION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. Load datasets
# ------------------------------------------------------------

air_quality = pd.read_csv("datasets/air_quality.csv")
weather = pd.read_csv("datasets/weather_history.csv")
energy = pd.read_csv("datasets/Energy_consumption.csv")

# ------------------------------------------------------------
# 2. Store datasets together for easy inspection
# ------------------------------------------------------------

datasets = {
    "Air Quality": air_quality,
    "Weather": weather,          
    "Energy": energy
}

# ------------------------------------------------------------
# 3. Basic dataset information
# ------------------------------------------------------------

for name, df in datasets.items():

    print("\n" + "=" * 70)
    print(f"{name.upper()} DATASET")
    print("=" * 70)

    print("\nShape:")
    print(df.shape)

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nFirst 5 Rows:")
    display(df.head())

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nDescriptive Statistics:")
    display(df.describe(include="all").T)
```

```python
# ============================================================
# DAV PROJECT
# STEP 2: DATA CLEANING & PREPROCESSING
# ============================================================

# ------------------------------------------------------------
# 1. Create copies so original datasets remain unchanged
# ------------------------------------------------------------

air_quality_clean = air_quality.copy()
weather_clean = weather.copy()
energy_clean = energy.copy()


# ------------------------------------------------------------
# 2. AIR QUALITY DATA CLEANING
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("AIR QUALITY - DATA CLEANING")
print("=" * 70)

# Check invalid negative pollutant values
pollutant_columns = [
    "PM2.5", "PM10", "NOx", "NO2",
    "SO2", "VOCs", "CO", "CO2", "CH4"
]

print("\nNegative values before cleaning:")
print((air_quality_clean[pollutant_columns] < 0).sum())

# Replace physically invalid negative values with NaN
air_quality_clean[pollutant_columns] = (
    air_quality_clean[pollutant_columns]
    .mask(air_quality_clean[pollutant_columns] < 0)
)

# Fill invalid values using median
for col in pollutant_columns:
    air_quality_clean[col] = (
        air_quality_clean[col]
        .fillna(air_quality_clean[col].median())
    )

print("\nMissing values after cleaning:")
print(air_quality_clean.isnull().sum())

print("\nShape after cleaning:")
print(air_quality_clean.shape)


# ------------------------------------------------------------
# 3. WEATHER DATA CLEANING
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("WEATHER - DATA CLEANING")
print("=" * 70)

# Convert date column to datetime
weather_clean["Formatted Date"] = pd.to_datetime(
    weather_clean["Formatted Date"],
    utc=True
)

# Fill missing precipitation type using mode
weather_clean["Precip Type"] = weather_clean["Precip Type"].fillna(
    weather_clean["Precip Type"].mode()[0]
)

# Remove duplicate rows
weather_clean = weather_clean.drop_duplicates()

# Remove constant column
if weather_clean["Loud Cover"].nunique() <= 1:
    weather_clean.drop(columns=["Loud Cover"], inplace=True)

print("\nMissing values after cleaning:")
print(weather_clean.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(weather_clean.duplicated().sum())

print("\nShape after cleaning:")
print(weather_clean.shape)


# ------------------------------------------------------------
# 4. ENERGY DATA CLEANING
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ENERGY - DATA CLEANING")
print("=" * 70)

# Convert timestamp to datetime
energy_clean["Timestamp"] = pd.to_datetime(
    energy_clean["Timestamp"]
)

print("\nMissing values after cleaning:")
print(energy_clean.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(energy_clean.duplicated().sum())

print("\nShape after cleaning:")
print(energy_clean.shape)


# ------------------------------------------------------------
# 5. Store cleaned datasets
# ------------------------------------------------------------

cleaned_datasets = {
    "Air Quality": air_quality_clean,
    "Weather": weather_clean,
    "Energy": energy_clean
}


# ------------------------------------------------------------
# 6. Final cleaning summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL CLEANING SUMMARY")
print("=" * 70)

for name, df in cleaned_datasets.items():

    print(f"\n{name}")
    print("-" * 40)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Missing Values:", df.isnull().sum().sum())
    print("Duplicate Rows:", df.duplicated().sum())

# ------------------------------------------------------------
# 7. SUMMARY STATISTICS OF CLEANED DATASETS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SUMMARY STATISTICS OF CLEANED DATASETS")
print("=" * 70)

for name, df in cleaned_datasets.items():

    print("\n" + "-" * 70)
    print(f"{name.upper()} - NUMERICAL SUMMARY")
    print("-" * 70)

    # Select numerical columns
    numerical_df = df.select_dtypes(include=np.number)

    display(
        numerical_df.describe().T.round(2)
    )


# ------------------------------------------------------------
# 8. CATEGORICAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CATEGORICAL DATA SUMMARY")
print("=" * 70)

for name, df in cleaned_datasets.items():

    print("\n" + "-" * 70)
    print(f"{name.upper()} - CATEGORICAL COLUMNS")
    print("-" * 70)

    categorical_columns = df.select_dtypes(
        include="object"
    ).columns

    for col in categorical_columns:

        print(f"\n{col}")
        print(df[col].value_counts())

# ------------------------------------------------------------
# 9. SAVE CLEANED DATASETS
# ------------------------------------------------------------

air_quality_clean.to_csv(
    "datasets/air_quality_cleaned.csv",
    index=False
)

weather_clean.to_csv(
    "datasets/weather_history_cleaned.csv",
    index=False
)

energy_clean.to_csv(
    "datasets/Energy_consumption_cleaned.csv",
    index=False
)

print("\nCleaned datasets saved successfully!")
```

```python
# ============================================================
# STRONGEST CORRELATIONS
# ============================================================

def strongest_correlations(corr_matrix, n=10):

    corr_pairs = (
        corr_matrix
        .where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
        .stack()
        .sort_values(key=abs, ascending=False)
    )

    return corr_pairs.head(n)


print("\n" + "=" * 70)
print("AIR QUALITY - STRONGEST CORRELATIONS")
print("=" * 70)
display(strongest_correlations(air_corr))

print("\n" + "=" * 70)
print("WEATHER - STRONGEST CORRELATIONS")
print("=" * 70)
display(strongest_correlations(weather_corr))

print("\n" + "=" * 70)
print("ENERGY - STRONGEST CORRELATIONS")
print("=" * 70)
display(strongest_correlations(energy_corr))
```

```python
# ============================================================
# STEP 4: FEATURE ENGINEERING
# 4.1 AIR QUALITY DATASET
# ============================================================

from sklearn.preprocessing import StandardScaler

# Make a copy of the cleaned dataset
air_fe = air_quality_clean.copy()

# ------------------------------------------------------------
# 1. PM2.5 / PM10 Ratio
# ------------------------------------------------------------

air_fe["PM_Ratio"] = (
    air_fe["PM2.5"] / air_fe["PM10"]
)

# ------------------------------------------------------------
# 2. NOx / NO2 Ratio
# ------------------------------------------------------------

air_fe["NOx_NO2_Ratio"] = (
    air_fe["NOx"] / air_fe["NO2"]
)

# ------------------------------------------------------------
# 3. CO / CO2 Ratio
# ------------------------------------------------------------

air_fe["CO_CO2_Ratio"] = (
    air_fe["CO"] / air_fe["CO2"]
)

# ------------------------------------------------------------
# 4. Pollution Score
# ------------------------------------------------------------

pollutants = [
    "PM2.5",
    "PM10",
    "NOx",
    "NO2",
    "SO2",
    "VOCs",
    "CO",
    "CO2",
    "CH4"
]

scaler = StandardScaler()

pollutant_scaled = scaler.fit_transform(
    air_fe[pollutants]
)

air_fe["Pollution_Score"] = (
    pollutant_scaled.mean(axis=1)
)

# ------------------------------------------------------------
# 5. Pollution Level
# ------------------------------------------------------------

air_fe["Pollution_Level"] = pd.qcut(
    air_fe["Pollution_Score"],
    q=3,
    labels=["Low", "Medium", "High"]
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("AIR QUALITY - FEATURE ENGINEERING")
print("=" * 70)

print("\nNew Features Created:")
print([
    "PM_Ratio",
    "NOx_NO2_Ratio",
    "CO_CO2_Ratio",
    "Pollution_Score",
    "Pollution_Level"
])

print("\nOriginal Shape:")
print(air_quality_clean.shape)

print("\nUpdated Shape:")
print(air_fe.shape)

print("\nFirst 5 Rows:")
display(
    air_fe[
        [
            "PM2.5",
            "PM10",
            "NOx",
            "NO2",
            "CO",
            "CO2",
            "PM_Ratio",
            "NOx_NO2_Ratio",
            "CO_CO2_Ratio",
            "Pollution_Score",
            "Pollution_Level"
        ]
    ].head()
)

# ============================================================
# VALIDATION OF ENGINEERED FEATURES
# ============================================================

new_features = [
    "PM_Ratio",
    "NOx_NO2_Ratio",
    "CO_CO2_Ratio",
    "Pollution_Score",
    "Pollution_Level"
]

print("=" * 70)
print("FEATURE ENGINEERING VALIDATION")
print("=" * 70)

print("\nMissing Values:")
print(
    air_fe[new_features].isnull().sum()
)

print("\nPollution Level Distribution:")
print(
    air_fe["Pollution_Level"].value_counts()
)

print("\nPollution Score Statistics:")
display(
    air_fe["Pollution_Score"].describe().round(3)
)

# ============================================================
# POLLUTION LEVEL DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=air_fe,
    x="Pollution_Level",
    order=["Low", "Medium", "High"]
)

plt.title("Distribution of Pollution Levels")
plt.xlabel("Pollution Level")
plt.ylabel("Number of Observations")

plt.tight_layout()
plt.show()

# ============================================================
# SAVE FEATURE-ENGINEERED AIR QUALITY DATASET
# ============================================================

air_fe.to_csv(
    "datasets/air_quality_feature_engineered.csv",
    index=False
)

print("Air Quality feature-engineered dataset saved successfully!")
```

```python
# ============================================================
# STEP 4.2 — WEATHER FEATURE ENGINEERING
# ============================================================

# Make a copy of the cleaned Weather dataset
weather_fe = weather_clean.copy()

# ------------------------------------------------------------
# Convert Formatted Date to datetime
# ------------------------------------------------------------

weather_fe["Formatted Date"] = pd.to_datetime(
    weather_fe["Formatted Date"],
    utc=True
)

# ------------------------------------------------------------
# 1. Year
# ------------------------------------------------------------

weather_fe["Year"] = (
    weather_fe["Formatted Date"].dt.year
)

# ------------------------------------------------------------
# 2. Month
# ------------------------------------------------------------

weather_fe["Month"] = (
    weather_fe["Formatted Date"].dt.month
)

# ------------------------------------------------------------
# 3. Hour
# ------------------------------------------------------------

weather_fe["Hour"] = (
    weather_fe["Formatted Date"].dt.hour
)

# ------------------------------------------------------------
# 4. Day of Week
# ------------------------------------------------------------

weather_fe["DayOfWeek"] = (
    weather_fe["Formatted Date"].dt.day_name()
)

# ------------------------------------------------------------
# 5. Season
# ------------------------------------------------------------

def get_season(month):

    if month in [12, 1, 2]:
        return "Winter"

    elif month in [3, 4, 5]:
        return "Spring"

    elif month in [6, 7, 8]:
        return "Summer"

    else:
        return "Autumn"


weather_fe["Season"] = (
    weather_fe["Month"].apply(get_season)
)

# ------------------------------------------------------------
# 6. Temperature Difference
# ------------------------------------------------------------

weather_fe["Temperature_Difference"] = (
    weather_fe["Temperature (C)"]
    - weather_fe["Apparent Temperature (C)"]
)

# ------------------------------------------------------------
# 7. Wind Chill Indicator
# ------------------------------------------------------------

weather_fe["Wind_Chill_Indicator"] = np.where(
    weather_fe["Apparent Temperature (C)"]
    < weather_fe["Temperature (C)"],
    "Yes",
    "No"
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("WEATHER - FEATURE ENGINEERING")
print("=" * 70)

print("\nNew Features Created:")

print([
    "Year",
    "Month",
    "Hour",
    "DayOfWeek",
    "Season",
    "Temperature_Difference",
    "Wind_Chill_Indicator"
])

print("\nOriginal Shape:")
print(weather_clean.shape)

print("\nUpdated Shape:")
print(weather_fe.shape)

print("\nFirst 5 Rows:")

display(
    weather_fe[
        [
            "Formatted Date",
            "Temperature (C)",
            "Apparent Temperature (C)",
            "Year",
            "Month",
            "Hour",
            "DayOfWeek",
            "Season",
            "Temperature_Difference",
            "Wind_Chill_Indicator"
        ]
    ].head()
)

# ============================================================
# WEATHER FEATURE VALIDATION
# ============================================================

weather_new_features = [
    "Year",
    "Month",
    "Hour",
    "DayOfWeek",
    "Season",
    "Temperature_Difference",
    "Wind_Chill_Indicator"
]

print("=" * 70)
print("WEATHER - FEATURE ENGINEERING VALIDATION")
print("=" * 70)

print("\nMissing Values:")
print(
    weather_fe[weather_new_features].isnull().sum()
)

print("\nSeason Distribution:")
print(
    weather_fe["Season"].value_counts()
)

print("\nDay of Week Distribution:")
print(
    weather_fe["DayOfWeek"].value_counts()
)

print("\nWind Chill Indicator:")
print(
    weather_fe["Wind_Chill_Indicator"].value_counts()
)

print("\nTemperature Difference Statistics:")
display(
    weather_fe["Temperature_Difference"]
    .describe()
    .round(3)
)

# ============================================================
# SEASONAL TEMPERATURE ANALYSIS
# ============================================================

season_order = [
    "Winter",
    "Spring",
    "Summer",
    "Autumn"
]

season_temperature = (
    weather_fe
    .groupby("Season")["Temperature (C)"]
    .mean()
    .reindex(season_order)
)

plt.figure(figsize=(8, 5))

season_temperature.plot(
    kind="bar"
)

plt.title("Average Temperature by Season")
plt.xlabel("Season")
plt.ylabel("Average Temperature (°C)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()

# ============================================================
# HOURLY TEMPERATURE PATTERN
# ============================================================

hourly_temperature = (
    weather_fe
    .groupby("Hour")["Temperature (C)"]
    .mean()
)

plt.figure(figsize=(10, 5))

hourly_temperature.plot(
    kind="line",
    marker="o"
)

plt.title("Average Temperature by Hour of Day")
plt.xlabel("Hour of Day")
plt.ylabel("Average Temperature (°C)")
plt.xticks(range(0, 24))

plt.tight_layout()
plt.show()

# ============================================================
# SAVE WEATHER FEATURE-ENGINEERED DATASET
# ============================================================

weather_fe.to_csv(
    "datasets/weather_feature_engineered.csv",
    index=False
)

print("Weather feature-engineered dataset saved successfully!")
```

```python
# ============================================================
# STEP 4.3 — ENERGY FEATURE ENGINEERING
# ============================================================

# Make a copy of the cleaned Energy dataset
energy_fe = energy_clean.copy()

# ------------------------------------------------------------
# Convert Timestamp to datetime
# ------------------------------------------------------------

energy_fe["Timestamp"] = pd.to_datetime(
    energy_fe["Timestamp"]
)

# ------------------------------------------------------------
# 1. Hour
# ------------------------------------------------------------

energy_fe["Hour"] = (
    energy_fe["Timestamp"].dt.hour
)

# ------------------------------------------------------------
# 2. Month
# ------------------------------------------------------------

energy_fe["Month"] = (
    energy_fe["Timestamp"].dt.month
)

# ------------------------------------------------------------
# 3. Weekend
# ------------------------------------------------------------

energy_fe["Weekend"] = np.where(
    energy_fe["DayOfWeek"].isin(
        ["Saturday", "Sunday"]
    ),
    "Yes",
    "No"
)

# ------------------------------------------------------------
# 4. HVAC Status
# ------------------------------------------------------------

energy_fe["HVAC_Status"] = np.where(
    energy_fe["HVACUsage"] == "On",
    1,
    0
)

# ------------------------------------------------------------
# 5. Lighting Status
# ------------------------------------------------------------

energy_fe["Lighting_Status"] = np.where(
    energy_fe["LightingUsage"] == "On",
    1,
    0
)

# ------------------------------------------------------------
# 6. Energy Per Occupant
# ------------------------------------------------------------

energy_fe["Energy_Per_Occupant"] = np.where(
    energy_fe["Occupancy"] > 0,
    energy_fe["EnergyConsumption"] /
    energy_fe["Occupancy"],
    0
)

# ------------------------------------------------------------
# 7. Renewable Energy Ratio
# ------------------------------------------------------------

energy_fe["Renewable_Energy_Ratio"] = np.where(
    energy_fe["EnergyConsumption"] > 0,
    energy_fe["RenewableEnergy"] /
    energy_fe["EnergyConsumption"],
    0
)

# ------------------------------------------------------------
# 8. Energy Per Square Foot
# ------------------------------------------------------------

energy_fe["Energy_Per_SquareFoot"] = (
    energy_fe["EnergyConsumption"] /
    energy_fe["SquareFootage"]
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("ENERGY - FEATURE ENGINEERING")
print("=" * 70)

print("\nNew Features Created:")

print([
    "Hour",
    "Month",
    "Weekend",
    "HVAC_Status",
    "Lighting_Status",
    "Energy_Per_Occupant",
    "Renewable_Energy_Ratio",
    "Energy_Per_SquareFoot"
])

print("\nOriginal Shape:")
print(energy_clean.shape)

print("\nUpdated Shape:")
print(energy_fe.shape)

print("\nFirst 5 Rows:")

display(
    energy_fe[
        [
            "Timestamp",
            "Occupancy",
            "EnergyConsumption",
            "RenewableEnergy",
            "Hour",
            "Month",
            "Weekend",
            "HVAC_Status",
            "Lighting_Status",
            "Energy_Per_Occupant",
            "Renewable_Energy_Ratio",
            "Energy_Per_SquareFoot"
        ]
    ].head()
)

# ============================================================
# ENERGY FEATURE VALIDATION
# ============================================================

energy_new_features = [
    "Hour",
    "Month",
    "Weekend",
    "HVAC_Status",
    "Lighting_Status",
    "Energy_Per_Occupant",
    "Renewable_Energy_Ratio",
    "Energy_Per_SquareFoot"
]

print("=" * 70)
print("ENERGY - FEATURE ENGINEERING VALIDATION")
print("=" * 70)

print("\nMissing Values:")
print(
    energy_fe[energy_new_features].isnull().sum()
)

print("\nWeekend Distribution:")
print(
    energy_fe["Weekend"].value_counts()
)

print("\nHVAC Status:")
print(
    energy_fe["HVAC_Status"].value_counts()
)

print("\nLighting Status:")
print(
    energy_fe["Lighting_Status"].value_counts()
)

print("\nEngineered Numerical Features:")
display(
    energy_fe[
        [
            "Energy_Per_Occupant",
            "Renewable_Energy_Ratio",
            "Energy_Per_SquareFoot"
        ]
    ].describe().round(3)
)

# ============================================================
# RENEWABLE ENERGY RATIO
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    energy_fe["Renewable_Energy_Ratio"],
    bins=30,
    kde=True
)

plt.title("Distribution of Renewable Energy Ratio")
plt.xlabel("Renewable Energy Ratio")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

# ============================================================
# ENERGY PER SQUARE FOOT
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=energy_fe,
    x="SquareFootage",
    y="Energy_Per_SquareFoot",
    hue="HVACUsage"
)

plt.title("Energy Efficiency vs Square Footage")
plt.xlabel("Square Footage")
plt.ylabel("Energy Consumption per Square Foot")

plt.tight_layout()
plt.show()

# ============================================================
# HOURLY ENERGY CONSUMPTION
# ============================================================

hourly_energy = (
    energy_fe
    .groupby("Hour")["EnergyConsumption"]
    .mean()
)

plt.figure(figsize=(10, 5))

hourly_energy.plot(
    kind="line",
    marker="o"
)

plt.title("Average Energy Consumption by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average Energy Consumption")
plt.xticks(range(0, 24))

plt.tight_layout()
plt.show()

# ============================================================
# SAVE FEATURE-ENGINEERED ENERGY DATASET
# ============================================================

energy_fe.to_csv(
    "datasets/energy_feature_engineered.csv",
    index=False
)

print("Energy feature-engineered dataset saved successfully!")
```

```python
# ============================================================
# STEP 5.3: LEAKAGE CHECK + FEATURE TYPE SEPARATION
# ============================================================

# ------------------------------------------------------------
# FUNCTION TO INSPECT FEATURES
# ------------------------------------------------------------

def inspect_features(X, y, dataset_name):

    print("\n" + "=" * 70)
    print(f"{dataset_name.upper()} - FEATURE INSPECTION")
    print("=" * 70)

    numerical_features = X.select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_features = X.select_dtypes(
        exclude=np.number
    ).columns.tolist()

    print("\nNumerical Features:")
    print(numerical_features)

    print("\nCategorical Features:")
    print(categorical_features)

    print("\nNumber of Numerical Features:")
    print(len(numerical_features))

    print("\nNumber of Categorical Features:")
    print(len(categorical_features))

    # --------------------------------------------------------
    # CORRELATION WITH TARGET
    # --------------------------------------------------------

    numeric_data = X.select_dtypes(include=np.number).copy()

    if len(numeric_data.columns) > 0:

        correlation = numeric_data.corrwith(y)

        correlation = correlation.sort_values(
            key=abs,
            ascending=False
        )

        print("\nCorrelation with Target:")
        display(correlation.round(3))


# ------------------------------------------------------------
# INSPECT ALL THREE DATASETS
# ------------------------------------------------------------

inspect_features(
    X_air,
    y_air,
    "Air Quality"
)

inspect_features(
    X_weather,
    y_weather,
    "Weather"
)

inspect_features(
    X_energy,
    y_energy,
    "Energy"
)
```

```python
# ============================================================
# STEP 5.4: PREPROCESSING PIPELINES
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer


# ------------------------------------------------------------
# FUNCTION TO CREATE PREPROCESSOR
# ------------------------------------------------------------

def create_preprocessor(X):

    numerical_features = X.select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_features = X.select_dtypes(
        exclude=np.number
    ).columns.tolist()

    # Numerical pipeline
    numerical_pipeline = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]
    )

    # Categorical pipeline
    categorical_pipeline = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ))
        ]
    )

    # Combine both pipelines
    preprocessor = ColumnTransformer(
        transformers=[
            ("numerical", numerical_pipeline, numerical_features),
            ("categorical", categorical_pipeline, categorical_features)
        ],
        remainder="drop"
    )

    return preprocessor


# ------------------------------------------------------------
# CREATE PREPROCESSORS
# ------------------------------------------------------------

air_preprocessor = create_preprocessor(X_air)

weather_preprocessor = create_preprocessor(X_weather)

energy_preprocessor = create_preprocessor(X_energy)


# ------------------------------------------------------------
# DISPLAY PREPROCESSORS
# ------------------------------------------------------------

print("=" * 70)
print("PREPROCESSING PIPELINES CREATED")
print("=" * 70)

print("\nAir Quality Preprocessor:")
print(air_preprocessor)

print("\nWeather Preprocessor:")
print(weather_preprocessor)

print("\nEnergy Preprocessor:")
print(energy_preprocessor)
```

```python
# ============================================================
# STEP 6.2: MODEL TRAINING
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# ============================================================
# 1. PREPARE AIR QUALITY DATA
# ============================================================

air_target = "Pollution_Score"

air_features = [
    "Temperature",
    "Humidity",
    "Wind_Direction",
    "Location_Type",
    "Source_Label"
]

X_air_model = air_selected[air_features].copy()
y_air_model = air_selected[air_target].copy()

# Train/Test Split
X_air_train, X_air_test, y_air_train, y_air_test = train_test_split(
    X_air_model,
    y_air_model,
    test_size=0.20,
    random_state=42
)


# ============================================================
# 2. PREPARE WEATHER DATA
# ============================================================

weather_target = "Temperature (C)"

weather_features = [
    "Summary",
    "Precip Type",
    "Humidity",
    "Wind Speed (km/h)",
    "Wind Bearing (degrees)",
    "Visibility (km)",
    "Pressure (millibars)",
    "Year",
    "Month",
    "Hour",
    "DayOfWeek",
    "Season"
]

X_weather_model = weather_selected[weather_features].copy()
y_weather_model = weather_selected[weather_target].copy()

# Chronological 80:20 split
split_index = int(len(X_weather_model) * 0.80)

X_weather_train = X_weather_model.iloc[:split_index].copy()
X_weather_test = X_weather_model.iloc[split_index:].copy()

y_weather_train = y_weather_model.iloc[:split_index].copy()
y_weather_test = y_weather_model.iloc[split_index:].copy()


# ============================================================
# 3. PREPARE ENERGY DATA
# ============================================================

energy_target = "EnergyConsumption"

energy_features = [
    "Temperature",
    "Humidity",
    "SquareFootage",
    "Occupancy",
    "HVACUsage",
    "LightingUsage",
    "RenewableEnergy",
    "DayOfWeek",
    "Holiday",
    "Hour",
    "Month",
    "Weekend",
    "HVAC_Status",
    "Lighting_Status"
]

X_energy_model = energy_selected[energy_features].copy()
y_energy_model = energy_selected[energy_target].copy()

# Train/Test Split
X_energy_train, X_energy_test, y_energy_train, y_energy_test = train_test_split(
    X_energy_model,
    y_energy_model,
    test_size=0.20,
    random_state=42
)


# ============================================================
# 4. CREATE PREPROCESSORS
# ============================================================

def create_model_preprocessor(X):

    numerical_features = X.select_dtypes(include=np.number).columns.tolist()
    categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

    numerical_pipeline = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]
    )

    categorical_pipeline = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ))
        ]
    )

    preprocessor = ColumnTransformer(
        transformers=[
            ("numerical", numerical_pipeline, numerical_features),
            ("categorical", categorical_pipeline, categorical_features)
        ],
        remainder="drop"
    )

    return preprocessor


air_model_preprocessor = create_model_preprocessor(X_air_train)
weather_model_preprocessor = create_model_preprocessor(X_weather_train)
energy_model_preprocessor = create_model_preprocessor(X_energy_train)


# ============================================================
# 5. PREPROCESS THE DATA
# ============================================================

X_air_train_processed = air_model_preprocessor.fit_transform(X_air_train)
X_air_test_processed = air_model_preprocessor.transform(X_air_test)

X_weather_train_processed = weather_model_preprocessor.fit_transform(X_weather_train)
X_weather_test_processed = weather_model_preprocessor.transform(X_weather_test)

X_energy_train_processed = energy_model_preprocessor.fit_transform(X_energy_train)
X_energy_test_processed = energy_model_preprocessor.transform(X_energy_test)


# ============================================================
# 6. TRAIN AIR QUALITY MODELS
# ============================================================

print("=" * 70)
print("AIR QUALITY MODEL TRAINING")
print("=" * 70)

air_models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
}

air_trained_models = {}

for name, model in air_models.items():

    model.fit(
        X_air_train_processed,
        y_air_train
    )

    air_trained_models[name] = model

    print(f"{name} trained successfully.")


# ============================================================
# 7. TRAIN WEATHER MODELS
# ============================================================

print("\n" + "=" * 70)
print("WEATHER MODEL TRAINING")
print("=" * 70)

weather_trained_models = {}

for name, model in weather_models.items():

    model.fit(
        X_weather_train_processed,
        y_weather_train
    )

    weather_trained_models[name] = model

    print(f"{name} trained successfully.")


# ============================================================
# 8. TRAIN ENERGY MODELS
# ============================================================

print("\n" + "=" * 70)
print("ENERGY MODEL TRAINING")
print("=" * 70)

energy_trained_models = {}

for name, model in energy_models.items():

    model.fit(
        X_energy_train_processed,
        y_energy_train
    )

    energy_trained_models[name] = model

    print(f"{name} trained successfully.")


# ============================================================
# 9. TRAINING SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("MODEL TRAINING COMPLETED")
print("=" * 70)

print("\nAir Quality Models:")
print(list(air_trained_models.keys()))

print("\nWeather Models:")
print(list(weather_trained_models.keys()))

print("\nEnergy Models:")
print(list(energy_trained_models.keys()))
```

```python
# ============================================================
# STEP 6.3: MODEL PREDICTION
# ============================================================

# ============================================================
# 1. AIR QUALITY PREDICTIONS
# ============================================================

print("=" * 70)
print("AIR QUALITY PREDICTIONS")
print("=" * 70)

air_predictions = {}

for name, model in air_trained_models.items():

    predictions = model.predict(
        X_air_test_processed
    )

    air_predictions[name] = predictions

    print(f"{name} predictions generated successfully.")
    print(f"Number of predictions: {len(predictions)}")


# ============================================================
# 2. WEATHER PREDICTIONS
# ============================================================

print("\n" + "=" * 70)
print("WEATHER PREDICTIONS")
print("=" * 70)

weather_predictions = {}

for name, model in weather_trained_models.items():

    predictions = model.predict(
        X_weather_test_processed
    )

    weather_predictions[name] = predictions

    print(f"{name} predictions generated successfully.")
    print(f"Number of predictions: {len(predictions)}")


# ============================================================
# 3. ENERGY PREDICTIONS
# ============================================================

print("\n" + "=" * 70)
print("ENERGY PREDICTIONS")
print("=" * 70)

energy_predictions = {}

for name, model in energy_trained_models.items():

    predictions = model.predict(
        X_energy_test_processed
    )

    energy_predictions[name] = predictions

    print(f"{name} predictions generated successfully.")
    print(f"Number of predictions: {len(predictions)}")


# ============================================================
# 4. DISPLAY SAMPLE PREDICTIONS
# ============================================================

print("\n" + "=" * 70)
print("SAMPLE PREDICTIONS")
print("=" * 70)

print("\nAir Quality - Random Forest:")
print(air_predictions["Random Forest"][:10])

print("\nWeather - Random Forest:")
print(weather_predictions["Random Forest"][:10])

print("\nEnergy - Random Forest:")
print(energy_predictions["Random Forest"][:10])


# ============================================================
# 5. PREDICTION SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PREDICTION COMPLETED")
print("=" * 70)

print(f"Air Quality predictions : {len(air_predictions['Random Forest'])}")
print(f"Weather predictions    : {len(weather_predictions['Random Forest'])}")
print(f"Energy predictions     : {len(energy_predictions['Random Forest'])}")
```

```python
# ============================================================
# STEP 7.2: FAST SHAP VALUE GENERATION
# ============================================================
print("=" * 70)
print("GENERATING SHAP VALUES")
print("=" * 70)

def generate_shap(model, X_data, dataset_name, sample_size):
    print(f"\n{dataset_name}")
    
    n_samples = min(sample_size, X_data.shape[0])
    X_sample = X_data[:n_samples]

    print("Samples used:", n_samples)
    print("Features:", X_sample.shape[1])

    if isinstance(model, (DecisionTreeRegressor, RandomForestRegressor)):
        explainer = shap.TreeExplainer(model)
        shap_values = explainer.shap_values(X_sample)

    elif isinstance(model, LinearRegression):
        background_size = min(20, X_data.shape[0])
        background = X_data[:background_size]

        explainer = shap.LinearExplainer(
            model,
            background
        )

        shap_values = explainer.shap_values(X_sample)

    else:
        raise ValueError("Unsupported model")

    print("SHAP completed.")
    print("SHAP shape:", np.array(shap_values).shape)

    return explainer, shap_values, X_sample


# Air Quality
air_explainer, air_shap_values, X_air_shap = generate_shap(
    final_air_model,
    X_air_test_processed,
    "Air Quality",
    100
)


# Weather
weather_explainer, weather_shap_values, X_weather_shap = generate_shap(
    final_weather_model,
    X_weather_test_processed,
    "Weather",
    50
)


# Energy
energy_explainer, energy_shap_values, X_energy_shap = generate_shap(
    final_energy_model,
    X_energy_test_processed,
    "Energy",
    100
)

print("\n" + "=" * 70)
print("ALL SHAP VALUES GENERATED SUCCESSFULLY")
print("=" * 70)
```

```python
# ==============================================================
# STEP 7.3 - SHAP FEATURE IMPORTANCE
# ==============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ==============================================================
# CREATE SHAP RESULT FOLDERS
# ==============================================================

air_shap_path = "SHAP_Results/Air_Quality"
weather_shap_path = "SHAP_Results/Weather"
energy_shap_path = "SHAP_Results/Energy"

os.makedirs(air_shap_path, exist_ok=True)
os.makedirs(weather_shap_path, exist_ok=True)
os.makedirs(energy_shap_path, exist_ok=True)


# ==============================================================
# SHAP FEATURE IMPORTANCE FUNCTION
# ==============================================================

def shap_feature_importance(
    shap_values,
    feature_names,
    dataset_name,
    save_path
):

    shap_values = np.array(shap_values)

    # Calculate mean absolute SHAP value
    mean_abs_shap = np.mean(
        np.abs(shap_values),
        axis=0
    )

    print("\n" + "=" * 70)
    print(f"{dataset_name.upper()} - SHAP FEATURE IMPORTANCE")
    print("=" * 70)

    print("SHAP features:", len(mean_abs_shap))
    print("Feature names:", len(feature_names))

    # Safety check
    if len(mean_abs_shap) != len(feature_names):
        raise ValueError(
            f"Feature mismatch for {dataset_name}: "
            f"SHAP has {len(mean_abs_shap)} features, "
            f"but feature names have {len(feature_names)}."
        )

    # Create importance dataframe
    importance_df = pd.DataFrame({
        "Feature": feature_names,
        "Mean_Absolute_SHAP": mean_abs_shap
    })

    # Sort by importance
    importance_df = importance_df.sort_values(
        by="Mean_Absolute_SHAP",
        ascending=False
    ).reset_index(drop=True)

    # Display
    display(importance_df)

    # ==========================================================
    # PLOT TOP 15 FEATURES
    # ==========================================================

    top_features = importance_df.head(15)

    plt.figure(figsize=(10, 7))

    plt.barh(
        top_features["Feature"][::-1],
        top_features["Mean_Absolute_SHAP"][::-1]
    )

    plt.xlabel("Mean Absolute SHAP Value")
    plt.ylabel("Feature")
    plt.title(
        f"SHAP Feature Importance - {dataset_name}"
    )

    plt.tight_layout()

    # Save graph
    graph_file = os.path.join(
        save_path,
        f"{dataset_name.replace(' ', '_')}_SHAP_Feature_Importance.jpeg"
    )

    plt.savefig(
        graph_file,
        format="jpeg",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    # ==========================================================
    # SAVE CSV
    # ==========================================================

    csv_file = os.path.join(
        save_path,
        f"{dataset_name.replace(' ', '_')}_SHAP_Importance.csv"
    )

    importance_df.to_csv(
        csv_file,
        index=False
    )

    print("\nGraph saved:", graph_file)
    print("CSV saved:", csv_file)

    return importance_df


# ==============================================================
# GET CORRECT FEATURE NAMES FROM MODEL PREPROCESSORS
# ==============================================================

air_feature_names = list(
    air_model_preprocessor.get_feature_names_out()
)

weather_feature_names = list(
    weather_model_preprocessor.get_feature_names_out()
)

energy_feature_names = list(
    energy_model_preprocessor.get_feature_names_out()
)


# ==============================================================
# AIR QUALITY
# ==============================================================

air_importance_df = shap_feature_importance(
    air_shap_values,
    air_feature_names,
    "Air Quality",
    air_shap_path
)


# ==============================================================
# WEATHER
# ==============================================================

weather_importance_df = shap_feature_importance(
    weather_shap_values,
    weather_feature_names,
    "Weather",
    weather_shap_path
)


# ==============================================================
# ENERGY
# ==============================================================

energy_importance_df = shap_feature_importance(
    energy_shap_values,
    energy_feature_names,
    "Energy",
    energy_shap_path
)


# ==============================================================
# COMPLETION
# ==============================================================

print("\n" + "=" * 70)
print("STEP 7.3 COMPLETED SUCCESSFULLY")
print("=" * 70)
```

```python
# ==============================================================
# STEP 7.4 - SHAP SUMMARY / BEESWARM PLOTS
# ==============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import shap


# ==============================================================
# OUTPUT FOLDERS
# ==============================================================

air_shap_path = "SHAP_Results/Air_Quality"
weather_shap_path = "SHAP_Results/Weather"
energy_shap_path = "SHAP_Results/Energy"

os.makedirs(air_shap_path, exist_ok=True)
os.makedirs(weather_shap_path, exist_ok=True)
os.makedirs(energy_shap_path, exist_ok=True)


# ==============================================================
# FUNCTION
# ==============================================================

def create_shap_summary(
    shap_values,
    X_sample,
    feature_names,
    dataset_name,
    save_path
):

    print("\n" + "=" * 70)
    print(f"{dataset_name.upper()} - SHAP SUMMARY PLOT")
    print("=" * 70)

    shap_values = np.asarray(shap_values)
    X_sample = np.asarray(X_sample)

    print("SHAP shape:", shap_values.shape)
    print("Data shape:", X_sample.shape)
    print("Features:", len(feature_names))

    # Safety check
    if shap_values.shape[1] != len(feature_names):
        raise ValueError(
            f"Feature mismatch for {dataset_name}: "
            f"SHAP has {shap_values.shape[1]} features, "
            f"but feature names have {len(feature_names)}."
        )

    # Create DataFrame for feature values
    X_sample_df = pd.DataFrame(
        X_sample,
        columns=feature_names
    )

    # Create SHAP summary plot directly
    shap.summary_plot(
        shap_values,
        X_sample_df,
        feature_names=feature_names,
        max_display=15,
        show=False
    )

    plt.title(
        f"SHAP Summary Plot - {dataset_name}"
    )

    plt.tight_layout()

    # Save
    graph_file = os.path.join(
        save_path,
        f"{dataset_name.replace(' ', '_')}_SHAP_Summary.jpeg"
    )

    plt.savefig(
        graph_file,
        format="jpeg",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print("Summary plot saved:", graph_file)


# ==============================================================
# AIR QUALITY
# ==============================================================

create_shap_summary(
    air_shap_values,
    X_air_shap,
    air_feature_names,
    "Air Quality",
    air_shap_path
)


# ==============================================================
# WEATHER
# ==============================================================

create_shap_summary(
    weather_shap_values,
    X_weather_shap,
    weather_feature_names,
    "Weather",
    weather_shap_path
)


# ==============================================================
# ENERGY
# ==============================================================

create_shap_summary(
    energy_shap_values,
    X_energy_shap,
    energy_feature_names,
    "Energy",
    energy_shap_path
)


# ==============================================================
# COMPLETION
# ==============================================================

print("\n" + "=" * 70)
print("STEP 7.4 COMPLETED SUCCESSFULLY")
print("=" * 70)
```

```python
# ============================================================
# STEP 9.1: AIR QUALITY - HYBRID RESIDUAL LEARNING MODEL
# ============================================================

import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 75)
print("AIR QUALITY - HYBRID RESIDUAL LEARNING MODEL (HRLM)")
print("=" * 75)


# ------------------------------------------------------------
# 1. BASELINE MODEL PREDICTIONS
# ------------------------------------------------------------

# Linear Regression was selected as the Air Quality baseline
# during the previous model comparison.

air_baseline_predictions = final_air_model.predict(
    X_air_test_processed
)

print("\n[1] Baseline predictions generated.")
print("Number of test samples:", len(air_baseline_predictions))


# ------------------------------------------------------------
# 2. CALCULATE TRAINING RESIDUALS
# ------------------------------------------------------------

# Generate predictions on the training data

air_train_predictions = final_air_model.predict(
    X_air_train_processed
)

# Residual = Actual - Predicted

air_train_residuals = (
    np.asarray(y_air_train)
    - air_train_predictions
)

print("\n[2] Training residuals calculated.")

print(
    "Residual mean:",
    round(np.mean(air_train_residuals), 6)
)

print(
    "Residual standard deviation:",
    round(np.std(air_train_residuals), 6)
)


# ------------------------------------------------------------
# 3. TRAIN THE RESIDUAL LEARNER
# ------------------------------------------------------------

# Random Forest learns the patterns present in the
# baseline model's prediction errors.

air_residual_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

air_residual_model.fit(
    X_air_train_processed,
    air_train_residuals
)

print("\n[3] Random Forest residual learner trained successfully.")


# ------------------------------------------------------------
# 4. PREDICT RESIDUALS ON TEST DATA
# ------------------------------------------------------------

air_predicted_residuals = air_residual_model.predict(
    X_air_test_processed
)

print("[4] Test residual predictions generated.")


# ------------------------------------------------------------
# 5. CREATE HRLM PREDICTION
# ------------------------------------------------------------

# Hybrid prediction =
# Baseline prediction + predicted residual

air_hrlm_predictions = (
    air_baseline_predictions
    + air_predicted_residuals
)

print("\n[5] HRLM predictions generated successfully.")


# ------------------------------------------------------------
# 6. METRIC CALCULATION FUNCTION
# ------------------------------------------------------------

def calculate_metrics(y_true, predictions):

    mae = mean_absolute_error(
        y_true,
        predictions
    )

    mse = mean_squared_error(
        y_true,
        predictions
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_true,
        predictions
    )

    return {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    }


# ------------------------------------------------------------
# 7. BASELINE PERFORMANCE
# ------------------------------------------------------------

air_baseline_metrics = calculate_metrics(
    y_air_test,
    air_baseline_predictions
)


# ------------------------------------------------------------
# 8. HRLM PERFORMANCE
# ------------------------------------------------------------

air_hrlm_metrics = calculate_metrics(
    y_air_test,
    air_hrlm_predictions
)


# ------------------------------------------------------------
# 9. BASELINE vs HRLM COMPARISON
# ------------------------------------------------------------

air_hrlm_comparison = pd.DataFrame(
    [
        {
            "Model": "Linear Regression (Baseline)",
            **air_baseline_metrics
        },
        {
            "Model": "HRLM (Linear Regression + Random Forest)",
            **air_hrlm_metrics
        }
    ]
)

print("\n")
print("=" * 75)
print("AIR QUALITY - BASELINE vs HRLM")
print("=" * 75)

print(
    air_hrlm_comparison.round(4)
)


# ------------------------------------------------------------
# 10. PERFORMANCE CHANGE
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("HRLM PERFORMANCE CHANGE")
print("=" * 75)

baseline_rmse = air_baseline_metrics["RMSE"]
hrlm_rmse = air_hrlm_metrics["RMSE"]

baseline_r2 = air_baseline_metrics["R2"]
hrlm_r2 = air_hrlm_metrics["R2"]


# RMSE improvement
rmse_change = (
    (baseline_rmse - hrlm_rmse)
    / baseline_rmse
) * 100


# R² improvement
r2_change = (
    (hrlm_r2 - baseline_r2)
    / abs(baseline_r2)
) * 100


print(
    f"RMSE change: {rmse_change:.2f}%"
)

print(
    f"R² change: {r2_change:.2f}%"
)


# ------------------------------------------------------------
# 11. RESULT SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("AIR QUALITY HRLM RESULT")
print("=" * 75)

if hrlm_rmse < baseline_rmse and hrlm_r2 > baseline_r2:

    print(
        "HRLM improved both RMSE and R² compared with "
        "the Linear Regression baseline."
    )

elif hrlm_rmse < baseline_rmse:

    print(
        "HRLM reduced RMSE compared with the baseline, "
        "but R² did not improve."
    )

elif hrlm_r2 > baseline_r2:

    print(
        "HRLM improved R² compared with the baseline, "
        "but RMSE did not improve."
    )

else:

    print(
        "HRLM did not improve the baseline on these metrics."
    )

print("=" * 75)
```

```python
# ============================================================
# STEP 9.1 (REVISED): AIR QUALITY HRLM WITH OOF RESIDUALS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 75)
print("AIR QUALITY - REVISED HYBRID RESIDUAL LEARNING MODEL (HRLM)")
print("=" * 75)


# ------------------------------------------------------------
# 1. PREPARE THE BASELINE MODEL
# ------------------------------------------------------------

# Linear Regression is the selected Air Quality baseline.

air_lr_model = LinearRegression()

print("\n[1] Linear Regression baseline initialized.")


# ------------------------------------------------------------
# 2. GENERATE OUT-OF-FOLD BASELINE PREDICTIONS
# ------------------------------------------------------------

# 5-fold cross-validation is used so that every training
# sample receives a prediction from a Linear Regression model
# that did NOT train on that sample.

air_kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

air_oof_predictions = cross_val_predict(
    air_lr_model,
    X_air_train_processed,
    y_air_train,
    cv=air_kfold,
    method="predict"
)

print("\n[2] Out-of-Fold baseline predictions generated.")
print(
    "Number of OOF predictions:",
    len(air_oof_predictions)
)


# ------------------------------------------------------------
# 3. CALCULATE OOF RESIDUALS
# ------------------------------------------------------------

# Residual = Actual - OOF Prediction

air_oof_residuals = (
    np.asarray(y_air_train)
    - air_oof_predictions
)

print("\n[3] Out-of-Fold residuals calculated.")

print(
    "OOF residual mean:",
    round(np.mean(air_oof_residuals), 6)
)

print(
    "OOF residual standard deviation:",
    round(np.std(air_oof_residuals), 6)
)


# ------------------------------------------------------------
# 4. TRAIN RESIDUAL LEARNER
# ------------------------------------------------------------

# Random Forest learns the patterns in the OOF residuals.

air_oof_residual_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

air_oof_residual_model.fit(
    X_air_train_processed,
    air_oof_residuals
)

print("\n[4] Random Forest residual learner trained successfully.")


# ------------------------------------------------------------
# 5. TRAIN FINAL LINEAR REGRESSION MODEL
# ------------------------------------------------------------

# After generating OOF residuals, train Linear Regression
# on the complete training set for final test prediction.

air_final_lr_model = LinearRegression()

air_final_lr_model.fit(
    X_air_train_processed,
    y_air_train
)

print("[5] Final Linear Regression model trained.")


# ------------------------------------------------------------
# 6. GENERATE TEST BASELINE PREDICTIONS
# ------------------------------------------------------------

air_oof_test_baseline_predictions = (
    air_final_lr_model.predict(
        X_air_test_processed
    )
)

print("[6] Test baseline predictions generated.")


# ------------------------------------------------------------
# 7. PREDICT TEST RESIDUALS
# ------------------------------------------------------------

air_oof_test_residual_predictions = (
    air_oof_residual_model.predict(
        X_air_test_processed
    )
)

print("[7] Test residual predictions generated.")


# ------------------------------------------------------------
# 8. CREATE FINAL HRLM PREDICTION
# ------------------------------------------------------------

air_oof_hrlm_predictions = (
    air_oof_test_baseline_predictions
    + air_oof_test_residual_predictions
)

print("\n[8] Revised HRLM predictions generated successfully.")


# ------------------------------------------------------------
# 9. METRIC FUNCTION
# ------------------------------------------------------------

def calculate_air_metrics(y_true, predictions):

    mae = mean_absolute_error(
        y_true,
        predictions
    )

    mse = mean_squared_error(
        y_true,
        predictions
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_true,
        predictions
    )

    return {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    }


# ------------------------------------------------------------
# 10. LINEAR REGRESSION BASELINE PERFORMANCE
# ------------------------------------------------------------

air_revised_baseline_metrics = calculate_air_metrics(
    y_air_test,
    air_oof_test_baseline_predictions
)


# ------------------------------------------------------------
# 11. REVISED HRLM PERFORMANCE
# ------------------------------------------------------------

air_revised_hrlm_metrics = calculate_air_metrics(
    y_air_test,
    air_oof_hrlm_predictions
)


# ------------------------------------------------------------
# 12. FINAL COMPARISON TABLE
# ------------------------------------------------------------

air_revised_hrlm_comparison = pd.DataFrame(
    [
        {
            "Model": "Linear Regression (Baseline)",
            **air_revised_baseline_metrics
        },
        {
            "Model": "Revised HRLM (LR + RF Residual)",
            **air_revised_hrlm_metrics
        }
    ]
)

print("\n")
print("=" * 75)
print("AIR QUALITY - REVISED BASELINE vs HRLM")
print("=" * 75)

print(
    air_revised_hrlm_comparison.round(4)
)


# ------------------------------------------------------------
# 13. PERFORMANCE CHANGE
# ------------------------------------------------------------

baseline_rmse = (
    air_revised_baseline_metrics["RMSE"]
)

hrlm_rmse = (
    air_revised_hrlm_metrics["RMSE"]
)

baseline_r2 = (
    air_revised_baseline_metrics["R2"]
)

hrlm_r2 = (
    air_revised_hrlm_metrics["R2"]
)


rmse_change = (
    (baseline_rmse - hrlm_rmse)
    / baseline_rmse
) * 100


r2_change = (
    (hrlm_r2 - baseline_r2)
    / abs(baseline_r2)
) * 100


print("\n")
print("=" * 75)
print("REVISED HRLM PERFORMANCE CHANGE")
print("=" * 75)

print(
    f"RMSE change: {rmse_change:.2f}%"
)

print(
    f"R² change: {r2_change:.2f}%"
)


# ------------------------------------------------------------
# 14. RESULT SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("AIR QUALITY - REVISED HRLM RESULT")
print("=" * 75)

if hrlm_rmse < baseline_rmse and hrlm_r2 > baseline_r2:

    print(
        "RESULT: Revised HRLM improved both RMSE and R²."
    )

elif hrlm_rmse < baseline_rmse:

    print(
        "RESULT: Revised HRLM reduced RMSE, "
        "but R² did not improve."
    )

elif hrlm_r2 > baseline_r2:

    print(
        "RESULT: Revised HRLM improved R², "
        "but RMSE did not improve."
    )

else:

    print(
        "RESULT: Revised HRLM did not improve "
        "the Linear Regression baseline."
    )

print("=" * 75)
```

```python
# ============================================================
# STEP 9.2: WEATHER - HYBRID RESIDUAL LEARNING MODEL
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import KFold, cross_val_predict
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 75)
print("WEATHER - HYBRID RESIDUAL LEARNING MODEL (HRLM)")
print("=" * 75)


# ------------------------------------------------------------
# 1. INITIALIZE BASELINE MODEL
# ------------------------------------------------------------

# Linear Regression is used as the baseline model.

weather_lr_model = LinearRegression()

print("\n[1] Linear Regression baseline initialized.")


# ------------------------------------------------------------
# 2. GENERATE OUT-OF-FOLD BASELINE PREDICTIONS
# ------------------------------------------------------------

# 5-fold cross-validation is used.
# Every training sample receives a prediction from a model
# that did not train on that particular sample.

weather_kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

weather_oof_predictions = cross_val_predict(
    weather_lr_model,
    X_weather_train_processed,
    y_weather_train,
    cv=weather_kfold,
    method="predict"
)

print("\n[2] Out-of-Fold baseline predictions generated.")

print(
    "Number of OOF predictions:",
    len(weather_oof_predictions)
)


# ------------------------------------------------------------
# 3. CALCULATE OOF RESIDUALS
# ------------------------------------------------------------

# Residual = Actual - OOF Prediction

weather_oof_residuals = (
    np.asarray(y_weather_train)
    - weather_oof_predictions
)

print("\n[3] Out-of-Fold residuals calculated.")

print(
    "OOF residual mean:",
    round(np.mean(weather_oof_residuals), 6)
)

print(
    "OOF residual standard deviation:",
    round(np.std(weather_oof_residuals), 6)
)


# ------------------------------------------------------------
# 4. TRAIN RESIDUAL LEARNER
# ------------------------------------------------------------

# Random Forest learns patterns in the residuals that
# Linear Regression could not capture.

weather_oof_residual_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

weather_oof_residual_model.fit(
    X_weather_train_processed,
    weather_oof_residuals
)

print(
    "\n[4] Random Forest residual learner "
    "trained successfully."
)


# ------------------------------------------------------------
# 5. TRAIN FINAL LINEAR REGRESSION MODEL
# ------------------------------------------------------------

# Train Linear Regression on the complete training set
# for final test-set baseline predictions.

weather_final_lr_model = LinearRegression()

weather_final_lr_model.fit(
    X_weather_train_processed,
    y_weather_train
)

print("[5] Final Linear Regression model trained.")


# ------------------------------------------------------------
# 6. GENERATE TEST BASELINE PREDICTIONS
# ------------------------------------------------------------

weather_test_baseline_predictions = (
    weather_final_lr_model.predict(
        X_weather_test_processed
    )
)

print("[6] Test baseline predictions generated.")


# ------------------------------------------------------------
# 7. PREDICT TEST RESIDUALS
# ------------------------------------------------------------

weather_test_residual_predictions = (
    weather_oof_residual_model.predict(
        X_weather_test_processed
    )
)

print("[7] Test residual predictions generated.")


# ------------------------------------------------------------
# 8. CREATE WEATHER HRLM PREDICTIONS
# ------------------------------------------------------------

# HRLM prediction =
# Linear Regression prediction + predicted residual

weather_hrlm_predictions = (
    weather_test_baseline_predictions
    + weather_test_residual_predictions
)

print(
    "\n[8] Weather HRLM predictions "
    "generated successfully."
)


# ------------------------------------------------------------
# 9. METRIC CALCULATION FUNCTION
# ------------------------------------------------------------

def calculate_weather_metrics(y_true, predictions):

    mae = mean_absolute_error(
        y_true,
        predictions
    )

    mse = mean_squared_error(
        y_true,
        predictions
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_true,
        predictions
    )

    return {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    }


# ------------------------------------------------------------
# 10. BASELINE PERFORMANCE
# ------------------------------------------------------------

weather_baseline_metrics = calculate_weather_metrics(
    y_weather_test,
    weather_test_baseline_predictions
)


# ------------------------------------------------------------
# 11. HRLM PERFORMANCE
# ------------------------------------------------------------

weather_hrlm_metrics = calculate_weather_metrics(
    y_weather_test,
    weather_hrlm_predictions
)


# ------------------------------------------------------------
# 12. BASELINE vs HRLM COMPARISON
# ------------------------------------------------------------

weather_hrlm_comparison = pd.DataFrame(
    [
        {
            "Model": "Linear Regression (Baseline)",
            **weather_baseline_metrics
        },
        {
            "Model": "HRLM (Linear Regression + Random Forest)",
            **weather_hrlm_metrics
        }
    ]
)

print("\n")
print("=" * 75)
print("WEATHER - BASELINE vs HRLM")
print("=" * 75)

print(
    weather_hrlm_comparison.round(4)
)


# ------------------------------------------------------------
# 13. PERFORMANCE CHANGE
# ------------------------------------------------------------

baseline_rmse = weather_baseline_metrics["RMSE"]
hrlm_rmse = weather_hrlm_metrics["RMSE"]

baseline_r2 = weather_baseline_metrics["R2"]
hrlm_r2 = weather_hrlm_metrics["R2"]


rmse_change = (
    (baseline_rmse - hrlm_rmse)
    / baseline_rmse
) * 100


r2_change = (
    (hrlm_r2 - baseline_r2)
    / abs(baseline_r2)
) * 100


print("\n")
print("=" * 75)
print("WEATHER HRLM PERFORMANCE CHANGE")
print("=" * 75)

print(
    f"RMSE change: {rmse_change:.2f}%"
)

print(
    f"R² change: {r2_change:.2f}%"
)


# ------------------------------------------------------------
# 14. RESULT SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("WEATHER HRLM RESULT")
print("=" * 75)

if hrlm_rmse < baseline_rmse and hrlm_r2 > baseline_r2:

    print(
        "RESULT: HRLM improved both RMSE and R²."
    )

elif hrlm_rmse < baseline_rmse:

    print(
        "RESULT: HRLM reduced RMSE, "
        "but R² did not improve."
    )

elif hrlm_r2 > baseline_r2:

    print(
        "RESULT: HRLM improved R², "
        "but RMSE did not improve."
    )

else:

    print(
        "RESULT: HRLM did not improve "
        "the Linear Regression baseline."
    )

print("=" * 75)
```

```python
# ============================================================
# 9.5 CROSS-DOMAIN FUSION
# Weather + Energy + Air Quality Context
# ============================================================

import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Load Feature-Selected Datasets
# ------------------------------------------------------------

air = pd.read_csv(
    "Feature_selected_datasets/air_quality_selected.csv"
)

weather = pd.read_csv(
    "Feature_selected_datasets/weather_selected.csv"
)

energy = pd.read_csv(
    "Feature_selected_datasets/energy_selected.csv"
)

print("Air Quality :", air.shape)
print("Weather     :", weather.shape)
print("Energy      :", energy.shape)


# ============================================================
# 2. Prepare Weather Dataset
# ============================================================

weather_fusion = weather[
    [
        "Month",
        "Hour",
        "Temperature (C)",
        "Humidity",
        "Wind Speed (km/h)",
        "Visibility (km)",
        "Pressure (millibars)",
        "Temperature_Difference",
        "Wind_Chill_Indicator"
    ]
].copy()

# Aggregate weather observations by Month + Hour
weather_fusion = (
    weather_fusion
    .groupby(["Month", "Hour"], as_index=False)
    .mean(numeric_only=True)
)

print("\nWeather Fusion Data:")
print(weather_fusion.head())

print("\nWeather Fusion Shape:")
print(weather_fusion.shape)


# ============================================================
# 3. Prepare Energy Dataset
# ============================================================

energy_fusion = energy[
    [
        "Timestamp",
        "Temperature",
        "Humidity",
        "SquareFootage",
        "Occupancy",
        "HVACUsage",
        "LightingUsage",
        "RenewableEnergy",
        "DayOfWeek",
        "Holiday",
        "EnergyConsumption",
        "Hour",
        "Month",
        "Weekend",
        "HVAC_Status",
        "Lighting_Status"
    ]
].copy()

print("\nEnergy Fusion Data:")
print(energy_fusion.head())


# ============================================================
# 4. Merge Weather + Energy
# ============================================================

fused_data = pd.merge(
    energy_fusion,
    weather_fusion,
    on=["Month", "Hour"],
    how="left"
)

print("\n========================================")
print("WEATHER + ENERGY FUSION")
print("========================================")

print("Shape:", fused_data.shape)

print("\nColumns:")
print(fused_data.columns.tolist())

print("\nMissing values:")
print(
    fused_data[
        [
            "Temperature (C)",
            "Humidity_y",
            "Wind Speed (km/h)",
            "Visibility (km)",
            "Pressure (millibars)"
        ]
    ].isnull().sum()
)


# ============================================================
# 5. Add Air Quality Context
# ============================================================

# Air Quality has no timestamp/month/hour.
# Therefore, temporal row-level fusion is not possible.
# We use its overall pollution characteristics as
# contextual smart-city indicators.

air_context = {
    "Air_PM25_Mean": air["PM2.5"].mean(),
    "Air_PM10_Mean": air["PM10"].mean(),
    "Air_NOx_Mean": air["NOx"].mean(),
    "Air_NO2_Mean": air["NO2"].mean(),
    "Air_SO2_Mean": air["SO2"].mean(),
    "Air_VOCs_Mean": air["VOCs"].mean(),
    "Air_CO2_Mean": air["CO2"].mean(),
    "Air_CO_Mean": air["CO"].mean(),
    "Air_CH4_Mean": air["CH4"].mean(),
    "Air_Humidity_Mean": air["Humidity"].mean(),
    "Air_Temperature_Mean": air["Temperature"].mean()
}

print("\nAir Quality Context:")
for key, value in air_context.items():
    print(f"{key}: {value:.4f}")


# Add Air Quality contextual features
for feature, value in air_context.items():
    fused_data[feature] = value


# ============================================================
# 6. Create Cross-Domain Interaction Features
# ============================================================

# Weather-Energy temperature difference
fused_data["Weather_Energy_Temp_Difference"] = (
    fused_data["Temperature (C)"]
    - fused_data["Temperature"]
)

# Weather-Energy humidity difference
fused_data["Weather_Energy_Humidity_Difference"] = (
    fused_data["Humidity_x"]
    - fused_data["Humidity_y"]
)

# Renewable energy contribution
fused_data["Renewable_Energy_Ratio"] = (
    fused_data["RenewableEnergy"]
    / (fused_data["EnergyConsumption"] + 1e-6)
)

# Occupancy relative to building size
fused_data["Occupancy_Density"] = (
    fused_data["Occupancy"]
    / (fused_data["SquareFootage"] + 1e-6)
)

print("\nCross-Domain Interaction Features:")
print(
    fused_data[
        [
            "Weather_Energy_Temp_Difference",
            "Weather_Energy_Humidity_Difference",
            "Renewable_Energy_Ratio",
            "Occupancy_Density"
        ]
    ].head()
)


# ============================================================
# 7. Handle Missing Values
# ============================================================

numeric_columns = fused_data.select_dtypes(
    include=np.number
).columns

fused_data[numeric_columns] = fused_data[numeric_columns].fillna(
    fused_data[numeric_columns].median()
)


# ============================================================
# 8. Save Cross-Domain Fusion Dataset
# ============================================================

output_folder = (
    "Feature_engineered_datasets/Cross_Domain_Fusion"
)

os.makedirs(output_folder, exist_ok=True)

fusion_path = os.path.join(
    output_folder,
    "cross_domain_fusion.csv"
)

fused_data.to_csv(
    fusion_path,
    index=False
)

print("\n========================================")
print("CROSS-DOMAIN FUSION COMPLETED")
print("========================================")

print("Final Shape:", fused_data.shape)

print("\nSaved to:")
print(fusion_path)


# ============================================================
# 9. Correlation Analysis
# ============================================================

correlation_columns = [
    "EnergyConsumption",
    "Temperature",
    "Humidity_x",
    "Temperature (C)",
    "Humidity_y",
    "Wind Speed (km/h)",
    "Visibility (km)",
    "Pressure (millibars)",
    "RenewableEnergy",
    "Occupancy",
    "Air_PM25_Mean",
    "Air_PM10_Mean",
    "Air_NOx_Mean",
    "Air_NO2_Mean"
]

correlation_columns = [
    col for col in correlation_columns
    if col in fused_data.columns
]

fusion_corr = fused_data[
    correlation_columns
].corr()

print("\nCross-Domain Correlation Matrix:")
print(fusion_corr["EnergyConsumption"].sort_values(
    ascending=False
))


# ============================================================
# 10. Plot Cross-Domain Correlations with Energy
# ============================================================

graph_folder = "EDA_Graphs/Cross_Domain_Fusion"

os.makedirs(graph_folder, exist_ok=True)

energy_corr = (
    fusion_corr["EnergyConsumption"]
    .drop("EnergyConsumption")
    .sort_values()
)

plt.figure(figsize=(10, 6))

bars = plt.barh(
    energy_corr.index,
    energy_corr.values
)

plt.axvline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "Cross-Domain Feature Relationships with Energy Consumption",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Correlation with Energy Consumption")

plt.ylabel("Cross-Domain Features")

for bar, value in zip(
    bars,
    energy_corr.values
):

    plt.text(
        value,
        bar.get_y() + bar.get_height() / 2,
        f" {value:.3f}",
        va="center",
        fontsize=9
    )

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()

correlation_graph = os.path.join(
    graph_folder,
    "01_Cross_Domain_Energy_Correlation.png"
)

plt.savefig(
    correlation_graph,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nGraph saved to:")
print(correlation_graph)
```

```python
# ============================================================
# 9.5.2 CROSS-DOMAIN FUSION MODEL
# Weather + Energy
# ============================================================

import pandas as pd
import numpy as np
import os

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. Load Fusion Dataset
# ============================================================

fusion_path = (
    "Feature_engineered_datasets/"
    "Cross_Domain_Fusion/"
    "cross_domain_fusion.csv"
)

fusion = pd.read_csv(fusion_path)

print("Fusion Dataset Shape:", fusion.shape)


# ============================================================
# 2. Define Target
# ============================================================

target = "EnergyConsumption"


# ============================================================
# 3. Select Cross-Domain Features
# ============================================================

features = [
    # -------------------------
    # Energy domain
    # -------------------------
    "Temperature",
    "Humidity_x",
    "SquareFootage",
    "Occupancy",
    "RenewableEnergy",
    "Hour",
    "Month",
    "HVAC_Status",
    "Lighting_Status",

    # -------------------------
    # Weather domain
    # -------------------------
    "Temperature (C)",
    "Humidity_y",
    "Wind Speed (km/h)",
    "Visibility (km)",
    "Pressure (millibars)",
    "Temperature_Difference",

    # -------------------------
    # Cross-domain features
    # -------------------------
    "Weather_Energy_Temp_Difference",
    "Weather_Energy_Humidity_Difference",
    "Renewable_Energy_Ratio",
    "Occupancy_Density"
]


# Keep only columns that actually exist
features = [
    feature
    for feature in features
    if feature in fusion.columns
]


X = fusion[features].copy()
y = fusion[target].copy()


print("\nNumber of Features:", len(features))

print("\nSelected Features:")
for feature in features:
    print("-", feature)


# ============================================================
# 4. Train-Test Split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Samples:", X_train.shape[0])
print("Testing Samples :", X_test.shape[0])


# ============================================================
# 5. Identify Numeric and Categorical Features
# ============================================================

categorical_features = []

numeric_features = X.columns.tolist()


# ============================================================
# 6. Preprocessing
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            "passthrough",
            numeric_features
        )
    ],
    remainder="drop"
)


# ============================================================
# 7. Define Models
# ============================================================

models = {

    "Fusion Linear Regression":
        LinearRegression(),

    "Fusion Decision Tree":
        DecisionTreeRegressor(
            random_state=42
        ),

    "Fusion Random Forest":
        RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        )
}


# ============================================================
# 8. Train and Evaluate Models
# ============================================================

fusion_results = []

trained_models = {}


for model_name, model in models.items():

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ]
    )

    pipeline.fit(
        X_train,
        y_train
    )

    predictions = pipeline.predict(
        X_test
    )

    mae = mean_absolute_error(
        y_test,
        predictions
    )

    mse = mean_squared_error(
        y_test,
        predictions
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_test,
        predictions
    )

    fusion_results.append({

        "Model": model_name,

        "MAE": mae,

        "MSE": mse,

        "RMSE": rmse,

        "R2": r2

    })

    trained_models[model_name] = pipeline


# ============================================================
# 9. Display Results
# ============================================================

fusion_results_df = pd.DataFrame(
    fusion_results
)

print("\n========================================")
print("CROSS-DOMAIN FUSION RESULTS")
print("========================================")

print(
    fusion_results_df.to_string(
        index=False
    )
)


# ============================================================
# 10. Compare with Existing Energy Models
# ============================================================

existing_energy_results = pd.DataFrame({

    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "MAE": [
        4.0909,
        5.8763,
        4.4574
    ],

    "MSE": [
        26.6679,
        54.3630,
        30.4221
    ],

    "RMSE": [
        5.1641,
        7.3731,
        5.5156
    ],

    "R2": [
        0.5929,
        0.1700,
        0.5355
    ]
})


# ============================================================
# 11. Add Fusion Results
# ============================================================

existing_energy_results["Type"] = "Existing"

fusion_results_df["Type"] = "Cross-Domain Fusion"

combined_results = pd.concat(
    [
        existing_energy_results,
        fusion_results_df
    ],
    ignore_index=True
)


print("\n========================================")
print("EXISTING vs CROSS-DOMAIN FUSION")
print("========================================")

print(
    combined_results.to_string(
        index=False
    )
)


# ============================================================
# 12. Save Results
# ============================================================

results_folder = (
    "Feature_engineered_datasets/"
    "Cross_Domain_Fusion"
)

os.makedirs(
    results_folder,
    exist_ok=True
)

results_path = os.path.join(
    results_folder,
    "cross_domain_fusion_model_results.csv"
)

combined_results.to_csv(
    results_path,
    index=False
)

print("\nResults saved to:")
print(results_path)
```

```python
# ============================================================
# 9.5.5 OVERALL CROSS-DOMAIN FINDINGS - CORRECTED
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# ------------------------------------------------------------
# 1. Output folders
# ------------------------------------------------------------

fusion_folder = "Feature_engineered_datasets/Cross_Domain_Fusion"
graph_folder = "EDA_Graphs/Cross_Domain_Fusion"

os.makedirs(fusion_folder, exist_ok=True)
os.makedirs(graph_folder, exist_ok=True)


# ============================================================
# 2. Load Weather + Energy Fusion Results
# ============================================================

fusion_results_path = os.path.join(
    fusion_folder,
    "cross_domain_fusion_model_results.csv"
)

fusion_results = pd.read_csv(fusion_results_path)

print("Weather + Energy Fusion Results:")
print(fusion_results.round(4))


# ============================================================
# 3. Separate Existing and Fusion Results
# ============================================================

existing_results = fusion_results[
    fusion_results["Type"] == "Existing"
].copy()

fusion_model_results = fusion_results[
    fusion_results["Type"] == "Cross-Domain Fusion"
].copy()


# ------------------------------------------------------------
# 4. Clean model names
# ------------------------------------------------------------

fusion_model_results["Base_Model"] = (
    fusion_model_results["Model"]
    .str.replace("Fusion ", "", regex=False)
)

existing_results["Base_Model"] = existing_results["Model"]


# ============================================================
# 5. Rename Metrics
# ============================================================

existing_results = existing_results.rename(
    columns={
        "RMSE": "Existing_RMSE",
        "R2": "Existing_R2"
    }
)

fusion_model_results = fusion_model_results.rename(
    columns={
        "RMSE": "Fusion_RMSE",
        "R2": "Fusion_R2"
    }
)


# ============================================================
# 6. Merge Existing vs Fusion Correctly
# ============================================================

overall_model_results = pd.merge(
    existing_results[
        [
            "Base_Model",
            "Existing_RMSE",
            "Existing_R2"
        ]
    ],
    fusion_model_results[
        [
            "Base_Model",
            "Fusion_RMSE",
            "Fusion_R2"
        ]
    ],
    on="Base_Model",
    how="inner"
)


# ============================================================
# 7. Calculate Improvements
# ============================================================

overall_model_results["RMSE_Improvement_%"] = (
    (
        overall_model_results["Existing_RMSE"]
        -
        overall_model_results["Fusion_RMSE"]
    )
    /
    overall_model_results["Existing_RMSE"]
) * 100


overall_model_results["R2_Improvement_%"] = (
    (
        overall_model_results["Fusion_R2"]
        -
        overall_model_results["Existing_R2"]
    )
    /
    overall_model_results["Existing_R2"]
) * 100


# ------------------------------------------------------------
# Rename for cleaner output
# ------------------------------------------------------------

overall_model_results = overall_model_results.rename(
    columns={
        "Base_Model": "Model"
    }
)


print("\n" + "=" * 60)
print("CORRECTED WEATHER + ENERGY MODEL IMPROVEMENT")
print("=" * 60)

print(
    overall_model_results.round(4)
)


# ============================================================
# 8. Save Corrected Results
# ============================================================

overall_model_path = os.path.join(
    fusion_folder,
    "overall_cross_domain_model_results.csv"
)

overall_model_results.to_csv(
    overall_model_path,
    index=False
)

print("\nSaved:")
print(overall_model_path)


# ============================================================
# 9. RMSE Improvement Graph
# ============================================================

plt.figure(figsize=(10, 6))

models = overall_model_results["Model"]

rmse_improvement = (
    overall_model_results["RMSE_Improvement_%"]
)

bars = plt.bar(
    models,
    rmse_improvement
)

plt.axhline(
    y=0,
    linewidth=1
)

plt.xlabel("Model")
plt.ylabel("RMSE Improvement (%)")

plt.title(
    "Weather + Energy Cross-Domain Fusion: RMSE Improvement"
)

for bar, value in zip(
    bars,
    rmse_improvement
):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.2f}%",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

rmse_graph = os.path.join(
    graph_folder,
    "07_Overall_Cross_Domain_RMSE_Improvement.png"
)

plt.savefig(
    rmse_graph,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", rmse_graph)


# ============================================================
# 10. R² Improvement Graph
# ============================================================

plt.figure(figsize=(10, 6))

r2_improvement = (
    overall_model_results["R2_Improvement_%"]
)

bars = plt.bar(
    models,
    r2_improvement
)

plt.axhline(
    y=0,
    linewidth=1
)

plt.xlabel("Model")
plt.ylabel("R² Improvement (%)")

plt.title(
    "Weather + Energy Cross-Domain Fusion: R² Improvement"
)

for bar, value in zip(
    bars,
    r2_improvement
):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.2f}%",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

r2_graph = os.path.join(
    graph_folder,
    "08_Overall_Cross_Domain_R2_Improvement.png"
)

plt.savefig(
    r2_graph,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", r2_graph)


# ============================================================
# 11. Load Air Quality + Weather Context
# ============================================================

air_weather_path = os.path.join(
    fusion_folder,
    "air_weather_environmental_comparison.csv"
)

air_weather = pd.read_csv(
    air_weather_path
)

print("\nAir Quality + Weather Context:")
print(
    air_weather.round(4)
)


# ============================================================
# 12. Final Cross-Domain Summary
# ============================================================

final_summary = pd.DataFrame({
    "Analysis": [
        "Weather + Energy Fusion",
        "Air Quality + Weather Context"
    ],

    "Method": [
        "Month-Hour Weather Context + Energy Prediction",
        "Environmental Distribution Comparison"
    ],

    "Purpose": [
        "Evaluate whether weather context improves energy consumption prediction",
        "Compare environmental conditions across Air Quality and Weather domains"
    ],

    "Result": [
        "Weather context improved Energy Consumption prediction across all three evaluated models",
        "Temperature and humidity profiles were compared after unit normalization"
    ]
})


print("\nFinal Cross-Domain Summary:")
print(final_summary)


# ============================================================
# 13. Save Final Summary
# ============================================================

summary_path = os.path.join(
    fusion_folder,
    "overall_cross_domain_summary.csv"
)

final_summary.to_csv(
    summary_path,
    index=False
)

print("\nSaved:")
print(summary_path)


# ============================================================
# 14. Environmental Context Graph
# ============================================================

environmental_data = air_weather.set_index(
    "Domain"
)

labels = [
    "Temperature",
    "Humidity"
]

x = np.arange(len(labels))

width = 0.35


air_values = [
    environmental_data.loc[
        "Air Quality",
        "Temperature_Mean"
    ],

    environmental_data.loc[
        "Air Quality",
        "Humidity_Mean_Percent"
    ]
]


weather_values = [
    environmental_data.loc[
        "Weather",
        "Temperature_Mean"
    ],

    environmental_data.loc[
        "Weather",
        "Humidity_Mean_Percent"
    ]
]


plt.figure(figsize=(10, 6))

plt.bar(
    x - width / 2,
    air_values,
    width,
    label="Air Quality"
)

plt.bar(
    x + width / 2,
    weather_values,
    width,
    label="Weather"
)

plt.xticks(
    x,
    labels
)

plt.ylabel("Value")
plt.xlabel("Environmental Variable")

plt.title(
    "Air Quality and Weather Environmental Context"
)

plt.legend()

plt.tight_layout()


environment_graph = os.path.join(
    graph_folder,
    "09_Air_Quality_Weather_Environmental_Context.png"
)

plt.savefig(
    environment_graph,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", environment_graph)


# ============================================================
# 15. Completion
# ============================================================

print("\n" + "=" * 60)
print("9.5 CROSS-DOMAIN FUSION COMPLETED CORRECTLY")
print("=" * 60)

print("\nCompleted:")
print("9.5.1 Dataset Alignment")
print("9.5.2 Weather + Energy Fusion Model")
print("9.5.3 Fusion Improvement Analysis")
print("9.5.4 Air Quality Cross-Domain Context Analysis")
print("9.5.5 Overall Cross-Domain Findings")
```

```python
# ============================================================
# STEP 15B — CREATE FEATURE ENGINEERING SUMMARY
# ============================================================

import os
import pandas as pd

# ------------------------------------------------------------
# 1. Dashboard artifact folder
# ------------------------------------------------------------

dashboard_root = "Dashboard_Artifacts"

feature_engineering_folder = os.path.join(
    dashboard_root,
    "feature_engineering"
)

os.makedirs(feature_engineering_folder, exist_ok=True)

# ------------------------------------------------------------
# 2. Define feature-engineered datasets
# ------------------------------------------------------------

datasets = {
    "Air Quality": air_feature_engineered,
    "Weather": weather_feature_engineered,
    "Energy": energy_feature_engineered
}

# ------------------------------------------------------------
# 3. Define target columns
# ------------------------------------------------------------

targets = {
    "Air Quality": "Pollution_Score",
    "Weather": "Temperature (C)",
    "Energy": "EnergyConsumption"
}

# ------------------------------------------------------------
# 4. Define engineered features
# ------------------------------------------------------------

engineered_features = {

    "Air Quality": [
        "PM_Ratio",
        "NOx_NO2_Ratio",
        "CO_CO2_Ratio"
    ],

    "Weather": [
        "Year",
        "Month",
        "Hour",
        "DayOfWeek",
        "Season",
        "Temperature_Difference",
        "Wind_Chill_Indicator"
    ],

    "Energy": [
        "Hour",
        "Month",
        "Weekend",
        "HVAC_Status",
        "Lighting_Status",
        "Energy_Per_Occupant",
        "Renewable_Energy_Ratio",
        "Energy_Per_SquareFoot"
    ]
}

# ------------------------------------------------------------
# 5. Create summary table
# ------------------------------------------------------------

summary_rows = []

for domain, df in datasets.items():

    target = targets[domain]

    for feature in engineered_features[domain]:

        if feature in df.columns:

            summary_rows.append({
                "Dataset": domain,
                "Engineered_Feature": feature,
                "Data_Type": str(df[feature].dtype),
                "Unique_Values": df[feature].nunique(),
                "Missing_Values": df[feature].isnull().sum(),
                "Target": target,
                "Rows": len(df)
            })

# Convert to DataFrame
feature_engineering_summary = pd.DataFrame(summary_rows)

# ------------------------------------------------------------
# 6. Display summary
# ------------------------------------------------------------

print("=" * 80)
print("FEATURE ENGINEERING SUMMARY")
print("=" * 80)

display(feature_engineering_summary)

# ------------------------------------------------------------
# 7. Save Excel and CSV
# ------------------------------------------------------------

excel_path = os.path.join(
    feature_engineering_folder,
    "feature_engineering_summary.xlsx"
)

csv_path = os.path.join(
    feature_engineering_folder,
    "feature_engineering_summary.csv"
)

feature_engineering_summary.to_excel(
    excel_path,
    index=False
)

feature_engineering_summary.to_csv(
    csv_path,
    index=False
)

print("\nFiles saved successfully:")
print(excel_path)
print(csv_path)
```
